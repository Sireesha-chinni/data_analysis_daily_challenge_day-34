Remove Duplicates & Preserve Order


Concepts: set(), list, loop
    
    
Problem:
    
    
Given a list of student IDs, remove duplicates without changing order.


ids = [101, 102, 101, 103, 102, 104]


Expected Output:
    
    
[101, 102, 103, 104]

```python
ids = [101,102,101,103,102,104]

num_set = set()

for i in ids:
    num_set.add(i)
    
num_set = list(num_set)

num_set.sort()


print(num_set)
```

Frequency Counter (Real-time analytics)


Concepts: dict, get(), loop
    
    
Problem:
    
    
Count how many times each product is sold.


products = ["apple", "banana", "apple", "orange", "banana", "apple"]


Expected Output:
    
    
    
{'apple': 3, 'banana': 2, 'orange': 1}

```python
products = ["apple", "banana", "apple", "orange", "banana", "apple"]

prot  = []

prot_cot = {}

for i in products:
    if i not in prot:
        prot.append(i)
        
for i in prot:
    prot_cot.update({i:products.count(i)})
    
print(prot_cot)
```

Find Top Scorer



Concepts: max(), dict
    
    
Problem:
    
    
Find the student with highest marks.


marks = {"A": 85, "B": 92, "C": 88}


Expected Output:
    
    
B

```python
marks = {"A": 85, "B": 92, "C": 88}

values_lit = []

values_lit.extend(marks.values())

values_lit.sort()

tag = values_lit[-1]

for key,value in marks.items():
    if value == tag:
        print(key)
```

Common Elements in Two Lists


Concepts: set(), intersection()
    
    
Problem:
    
    
Find common skills between two candidates.


skills1 = ["Python", "SQL", "Excel"]


skills2 = ["SQL", "Power BI", "Python"]


Expected Output:
    
    
['Python', 'SQL']

```python
skills1 = ["Python", "SQL", "Excel"]
skills2 = ["SQL", "Power BI", "Python"]

skills1 = set(skills1)
skills2 = set(skills2)

print(list(skills1.intersection(skills2)))
```

Sort Dictionary by Values


Concepts: sorted(), lambda
    
    
Problem:
    
    
Sort products by price.


prices = {"apple": 50, "banana": 20, "orange": 30}


Expected Output:
    
    
[('banana', 20), ('orange', 30), ('apple', 50)]

```python
prices = {"apple": 50, "banana": 20, "orange": 30}

pric = sorted(prices.items(), key = lambda item: item[1])

print(pric)
    
```
